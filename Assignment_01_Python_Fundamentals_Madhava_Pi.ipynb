I picked the Madhava pi example. The basic example uses a series to get closer and closer to pi. I changed it so the code tries a lot of different numbers of terms instead of only doing one calculation.

The main thing I wanted to check was: how many terms does it take before the answer is close enough?

## Method

The Madhava-Leibniz series looks like this:

$$\pi \approx 4 \sum_{n=0}^{N-1} \frac{(-1)^n}{2n+1}$$

Here, N is the number of terms I use. Since the series goes on forever, the computer only uses part of it. So the answer depends on how many terms I decide to add.

For each value of N, I calculated the error like this:

Absolute Error = |Calculated Value-Expected Value|


I used Python's `math.pi` as the expected value.

```python
import math
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8-whitegrid")
```

## Function

I put the pi calculation into a function so I could call it over and over with different numbers of terms.

```python
def madhava_pi(num_terms):
    """Approximate pi using num_terms of the Madhava-Leibniz series."""
    if num_terms <= 0:
        raise ValueError("num_terms must be a positive integer")

    total = 0.0
    for n in range(num_terms):
        sign = (-1) ** n
        denominator = 2 * n + 1
        total = total + sign / denominator

    return 4 * total
```

## Testing the Function

The assignment asks for at least three input values, so I tried a few different sizes.

```python
test_terms = [1, 10, 100, 1000]

for terms in test_terms:
    approximation = madhava_pi(terms)
    error = abs(approximation - math.pi)
    print(f"terms = {terms:4d} | pi approximation = {approximation:.12f} | absolute error = {error:.6e}")
```

The first answer is not very close to pi, but the answers get better when I use more terms.

## Experiment: Values and Errors

Next I ran the function for term counts from 1 to 5000. I saved the calculated values and the errors in lists so I could graph them.

```python
iteration_counts = list(range(1, 5001))
calculated_values = []
absolute_errors = []

for terms in iteration_counts:
    value = madhava_pi(terms)
    error = abs(value - math.pi)

    calculated_values.append(value)
    absolute_errors.append(error)

print(f"Stored {len(calculated_values)} calculated values.")
print(f"Final approximation with 5000 terms: {calculated_values[-1]:.12f}")
print(f"Final absolute error: {absolute_errors[-1]:.6e}")
```

## Figure 1: Getting Closer to pi

This graph shows the approximation as I use more terms. The dashed line is the real value of pi from `math.pi`.

```python
fig, ax = plt.subplots(figsize=(9, 5))

ax.plot(iteration_counts, calculated_values, color="#1f77b4", linewidth=1.0, label="Madhava approximation")
ax.axhline(math.pi, color="#d62728", linestyle="--", linewidth=1.5, label="Expected value: math.pi")

ax.set_title("Madhava Series Approximation of pi")
ax.set_xlabel("Number of terms")
ax.set_ylabel("Calculated value")
ax.legend()

plt.show()
```

## Figure 2: Error

This graph shows the absolute error. I used a log scale because the error gets smaller and is easier to see this way.

```python
fig, ax = plt.subplots(figsize=(9, 5))

ax.plot(iteration_counts, absolute_errors, color="#2ca02c", linewidth=1.0)
ax.set_yscale("log")

ax.set_title("Absolute Error of the Madhava Series")
ax.set_xlabel("Number of terms")
ax.set_ylabel("Absolute error, log scale")

plt.show()
```

## Reasonable Accuracy

I decided that reasonable accuracy means the absolute error is less than `0.001`. That means my answer is within one-thousandth of pi. This seemed close enough for this assignment without making the calculation too huge.

```python
reasonable_error = 0.001
terms_needed = None

for terms, error in zip(iteration_counts, absolute_errors):
    if error < reasonable_error:
        terms_needed = terms
        break

print(f"Reasonable accuracy threshold: absolute error < {reasonable_error}")
print(f"Number of terms required: {terms_needed}")
print(f"Approximation at that point: {madhava_pi(terms_needed):.12f}")
print(f"Absolute error at that point: {abs(madhava_pi(terms_needed) - math.pi):.6e}")
```
